```python
cal = Calendar()
cal["dtstart"] = "20260223T000000"
cal["dtend"] =   "20260903T235999"
cal["summary"] = f"schedu;e of the {group_id} group. "
cal["version"] = "2.0"
cal["PRODID"] = f"-//Apolo schedule export//{group_id}"
cal["Timezone"] = "CEST"
```

```python
for x, row in Classes.iterrows():
    event = Event()

    event["dtstart"] = row["Termin"].replace("-", "")+"T"+row["start time"].replace(":", "")+"00"
    event["dtend"] = row["Termin"].replace("-", "")+"T"+row["end time"].replace(":", "")+"00"
    event["location"] = row['Sala']
    event["Summary"] = row['Przedmiot']
    event['description'] = row["Nauczyciel"]
    event["UID"] = uuid.uuid4()
    event["DTSTAMP"] = vDatetime(datetime.datetime.now())
    
    cal.add_component(event)

    
```

```python
print(cal.to_ical())
```

```python
cf = open(f"Calanders/{group_id}.ics", "wb")
cf.write(cal.to_ical())
cf.close()
```
